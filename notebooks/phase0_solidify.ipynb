# Prime Engine — Phase 0, The Return

A deterministic house owns pose and props. Solidify stores each room. A small diffusion model renders frames conditioned on that cache.

The exit check: knock the foyer vase over, leave through three rooms, come back. The state hash matches, and the revisit is closer to the truth with Solidify than without it.

Colab already has PyTorch. A free T4 is enough. Raise `steps` if you want a longer fit. This notebook does not download third-party weights.

```python
import importlib.util
import os
import subprocess
import sys
from pathlib import Path

BRANCH = "cursor/phase0-solidify-prototype-2865"
if not Path("prime_engine").exists():
    subprocess.check_call([
        "git", "clone", "--branch", BRANCH, "--depth", "1",
        "https://github.com/AaronGrace978/Prime-Graphic-Ai-Engine.git",
        "prime-src",
    ])
    os.chdir("prime-src")
if importlib.util.find_spec("torch") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "-q", "torch",
        "--index-url", "https://download.pytorch.org/whl/cpu",
    ])
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"])
print("ready", Path.cwd())
```
